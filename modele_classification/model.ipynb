```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier

titanic= sns.load_dataset('titanic')


titanic= titanic[['survived', 'pclass', 'sex', 'age']]
titanic= titanic.dropna(axis=0)
titanic['sex'] = titanic['sex'].replace(['male', 'female'], [0, 1])
titanic.head()

model= KNeighborsClassifier()
y= titanic['survived']
X= titanic.drop('survived', axis=1)
model.fit(X,y)
model.score(X,y)

def ma_survie (model, pclass= 2, sex=1, age=20):
    x= np.array([pclass, sex, age]).reshape(1, 3)
    print(model.predict(x))
    survie= model.predict(x)
    if survie == 0:
        print("Vous n'aurez pas pu survivre au Titanic")
    else:
        print("Vous aurez survie au Titanic")
    print("Vos chance de ne pas survivre/ de survivre:")
    print(model.predict_proba(x))
ma_survie(model)


```
