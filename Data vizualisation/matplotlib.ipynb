```python
import pandas as pd
import matplotlib.pyplot as plt



#créer sa propre courbe
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot([1, 2, 3], [2, 4, 9], label="courbe", color="tab:blue", linestyle="--", marker="o")
ax.set_title("Mon titre")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

#Créer un graphique à partir des données en listes
jours = ["Lundi", "Mardi", "Mercredi", "Jeudi", "Vendredi"]
categories = ["Téléphones", "PC", "Tablettes", "Electroménager", "Micro"]
ventes = [100, 150, 120, 200, 250]
ages = [18, 20, 21, 22, 22, 25, 28, 30, 32, 35]


fig, ax = plt.subplots(figsize=(8, 5))

#Enlever le diez pour découvrir les autres graphiques

#ax.bar(categories, ventes, color="purple")
#ax.hist(ages, bins=5, color="skyblue", edgecolor="black")
#ax.scatter(jours, ventes, c="green", s=80)
#ax.pie(ventes, labels=categories, autopct="%1.1f%%", startangle=90)
ax.boxplot(ventes, vert= False, patch_artist=True, showmeans=True)


ax.set_title("Évolution des ventes")
ax.set_xlabel("Jours")
ax.set_ylabel("Ventes en milliers d'Ariary")

ax.legend()
ax.grid(alpha=0.5)

plt.tight_layout()
plt.show()

#Titanic : passager survivant et décédé
df1= pd.read_csv("titanic.csv")
resultat = df1["Survived"].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(8, 5))

ax.pie(
    resultat.values,
    labels=["Décédés", "Survivants"],
    autopct="%1.1f%%",
    startangle=90
)

ax.set_title("Répartition des passagers du Titanic selon leur survie")

plt.tight_layout()
plt.show()

#plusieurs graphiques d'un coup
df= pd.read_csv("titanic.csv")
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))
axes[0].hist(df["Age"], bins=20)
axes[1].boxplot(df["Age"].dropna())
fig.suptitle("Âge des passagers")
```
