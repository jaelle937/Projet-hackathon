```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv("titanic.csv")

sns.set_theme(style="whitegrid")

#histogramme
sns.countplot(data=df, x="Survived")

plt.title("Nombre de passagers selon leur survie")
plt.xlabel("Survie (0 = Décédé, 1 = Survivant)")
plt.ylabel("Nombre de passagers")

plt.show()

#barre
sns.barplot(data=df, x="Sex", y="Survived")

plt.title("Taux moyen de survie selon le sexe")
plt.xlabel("Sexe")
plt.ylabel("Taux de survie")

plt.show()

#distribution
sns.histplot(data=df, x="Age", bins=20)

plt.title("Distribution des âges des passagers")
plt.xlabel("Âge")
plt.ylabel("Nombre de passagers")

plt.show()

#boîte de moustache
sns.boxplot(data=df, x="Pclass", y="Fare")

plt.title("Prix des billets selon la classe")
plt.xlabel("Classe")
plt.ylabel("Prix du billet")

plt.show()

#corrélation entre variables - Heatmap
colonnes = ["Survived", "Pclass", "Age", "SibSp",
            "Parch", "Fare"]

correlation = df[colonnes].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matrice de corrélation du Titanic")
plt.show()

#analyse du heatmap - interprétation
print("                     Interprétation du Heatmap - Corrélation                   ")
print("Plus le numéro de classe augmente (de la 1re à la 3e classe), moins les passagers ont tendance à survivre.")
print("Les passagers ayant payé des billets plus chers ont généralement davantage survécu.")
print("L'âge présente peu de relation linéaire avec la survie des passagers.")
print("Le nombre de frères, sœurs ou conjoints à bord présente très peu de relation linéaire avec la survie.")
print("Le nombre de parents ou d'enfants à bord présente très peu de relation linéaire avec la survie.")

#KDE
sns.jointplot(
    data=df,
    x="Age",
    y="Fare",
    kind="kde",
    fill=True
)

plt.show()
```
