```python
import pandas as pd
import numpy as np

A= np.random.randint(0,10, [2,3])
print("Voici le tableau A :")
print(A)
B= np.sort(A) #trie les valeurs du plus grand au plus petit
print("Voici le tableau B :")
print(B)

#Compteur d'apparition de la valeur
value, counts = np.unique(A, return_counts=True)
for i, j in zip(value[counts.argsort()], counts[counts.argsort()]):
    print(f'la valeur {i} apparait {j} fois')

#A<10 #booléen indexing
#A[A<10]=10 #change les valeurs inférieures à 10 par 10
#print(A)
somme= np.sum(B) #calcul les sommes des valeurs du tableau
print("La somme du second tableau donne en tout :",somme)

#Calcule la coorélation entre les valeurs ligne par ligne 
# corrL1L1 = 1 , corrL1L2 != 1
coorelation= np.corrcoef(B)
print(coorelation)

#connaitre le pourcentage de données vides ou case vides dans le tableau
p_vide= np.isnan(B).sum()/B.size
print(f"Il y a {p_vide}% de données maquantes dans notre tableau")

```
