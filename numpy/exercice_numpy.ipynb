```python
## ✍️ Exercice 1 — NumPy

# 1. Crée un tableau `notes` de 20 entiers aléatoires entre 0 et 20 (graine fixée à 0).
# 2. Calcule moyenne, médiane, écart-type (échantillon), min, max.
# 3. Affiche uniquement les notes `>= 10`, puis leur nombre.
# 4. Remplace par `"Admis"` / `"Refusé"` avec `np.where`.
# 5. Transforme `notes` en matrice `(4, 5)` et calcule la moyenne de chaque colonne puis de chaque ligne.
# 6. **Standardise** le tableau : `(notes - moyenne) / écart-type`. Vérifie que la nouvelle moyenne ≈ 0 et l'écart-type = 1.
# 7. **Bonus** : sans boucle `for`, calcule la distance euclidienne entre deux vecteurs `a = [1, 2, 3]` et `b = [4, 6, 8]`.

import numpy as np

#1
notes= np.random.seed(0) #figé
notes= np.random.randint(0, 21, size=20)
print(notes)

#2
moyenne = np.mean(notes)
median =np.median(notes)
ecartype= np.std(notes, ddof=1)
print(f"La note maximale est de {notes.max()} tandisque la note minimale est de {notes.min()}")
print(f"La moyenne est de {moyenne}, la médiane est de {median} et l'écart-type est de {ecartype}")

#3
notes10= notes[notes >= 10]
#ce que j'ai fait mais sais trop compliqué
#compte= np.unique(notes10, return_counts=True)
#print(compte)
n10= len(notes10) # ou notes10.size c'est la même chose
print(f"Les notes supérieurs et égales à 10 sont au nombre de {n10} et les voilà : {notes10}")

#4
appreciation= np.where(notes >= 10, "Admis", "Refusé")
print(appreciation)

#5
transformation= notes.reshape(4,5)
print(transformation)
moyennecol= np.mean(transformation, axis=0)
print("La moyenne des notes par colonne est de :", moyennecol)
moyennelgn= np.mean(transformation, axis=1)
print("La moyenne des notes par ligne est de :", moyennelgn)

#6
standardisation= (notes - moyenne) / ecartype
print(standardisation)
print("La nouvelle moyenne est de :", np.mean(standardisation))
print("Le nouvel écart-type est de :", np.std(standardisation, ddof=1))

#7
a = [1, 2, 3]
b = [4, 6, 8]
distance_eucludienne= np.dot(a, b)
print("La distance euclidienne entre les deux vecteurs est de : ",distance_eucludienne)

```
