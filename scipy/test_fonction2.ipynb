```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.interpolate import interp1d
from scipy import optimize

#A exécuter un par un pour voir le résultat

#interpolate
x= np.linspace(0, 10,10)
y= x**2
plt.scatter(x, y)

f= interp1d(x, y, kind="linear")
new_x= np.linspace(0,10, 30)
result= f(new_x)
plt.scatter(new_x, result, c="r")


#optimize
x1= np.linspace(0, 2, 100)
y1= 1/3*x1**3-3/5*x1**2+2+ np.random.randn(x1.shape[0])/20
plt.scatter(x1, y1)
def f(x1, a, b, c, d):
    return a*x1**3+b*x1**2+c*x1+d
params, params_cov= optimize.curve_fit(f, x1, y1)
plt.plot(x1, f(x1, params[0], params[1], params[2], params[3]), c='g', linewidth=3)


#Optimie minimize
def f(x):
    return x**2+15* np.sin(x)
x= np.linspace(-10, 10, 100)
plt.plot(x, f(x))
optimize.minimize(f, x0= -8)
```
