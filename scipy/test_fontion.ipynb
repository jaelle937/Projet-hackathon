```python
import pandas as pd
import numpy as np

from scipy import stats

df = pd.read_csv("titanic.csv")

#Test de Chi2
tableau = pd.crosstab(df["Sex"], df["Survived"])

print(tableau)

chi2, p_value, dof, expected = stats.chi2_contingency(tableau)

print("Chi-deux :", chi2)
print("P-value :", p_value) 
if p_value < 0.05:
    print("H1 accepté")
else:
    print("H0 accepté")
print("Degrés de liberté :", dof)
print("Effectifs théoriques :")
print(expected)

#Test de Student 
survivants = df[df["Survived"] == 1]["Age"].dropna()

decedes = df[df["Survived"] == 0]["Age"].dropna()

t_stat, p_value = stats.ttest_ind(
    survivants,
    decedes,
    equal_var=False
)

print("Âge moyen des survivants :", survivants.mean())
print("Âge moyen des décédés :", decedes.mean())

print("Statistique T :", t_stat)
print("P-value :", p_value)
if p_value < 0.05:
    print("H1 accepté")
else:
    print("H0 accepté")

#La corrélation de Pearson
donnees = df[["Age", "Fare"]].dropna()

r, p_value = stats.pearsonr(
    donnees["Age"],
    donnees["Fare"]
)

print("Corrélation :", r)
print("P-value :", p_value)
if p_value < 0:
    print("Les deux variables évoluent dans le sens opposé.")
else:
    print("Les deux variables évoluent dans le même sens.")


#Test ANOVA
classe1 = df[df["Pclass"] == 1]["Age"].dropna()

classe2 = df[df["Pclass"] == 2]["Age"].dropna()

classe3 = df[df["Pclass"] == 3]["Age"].dropna()

f_stat, p_value = stats.f_oneway(
    classe1,
    classe2,
    classe3
)

print("Statistique F :", f_stat)
print("P-value :", p_value)
```
