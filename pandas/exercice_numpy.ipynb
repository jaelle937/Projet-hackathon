Crée (ou télécharge) un petit fichier `ventes.csv` avec : `date`, `produit`, `categorie`, `quantite`, `prix_unitaire`, `ville`. (Tu peux le générer avec NumPy.) Puis :

1. Charge-le, affiche `info()` et `describe()`.
2. Ajoute une colonne `chiffre_affaires = quantite * prix_unitaire`.
3. Quel est le chiffre d'affaires total par ville ? par catégorie ?
4. Quels sont les 3 produits les plus vendus (en quantité) ?
5. Convertis `date` en datetime, puis calcule le CA **par mois**.
6. Introduis volontairement 5 NaN dans `quantite` puis remplace-les par la médiane.
7. Crée un tableau croisé ville × catégorie (somme du CA).
8. Encode `categorie` en one-hot avec `get_dummies`.
9. **Bonus** : reproduis la question 3 en SQL (écris la requête) et compare avec ton résultat pandas.

```python
import pandas as pd
import numpy as np

df= pd.DataFrame({
    "date": ["30/09/2026", "30/09/2026", "10/10/2026", "11/10/2026", "12/10/2026", "13/10/2026"],
    "produit": ["Crayon", "Stylo", "Carnet", "Trousse", "Règle", "cahier"],
    "categorie": ["ustensile", "ustensile", "papeterie", "ustensile", "ustensile", "papeterie"],
    "prixu": [1000, 700, 15000, 8000, 3500, 5000],
    "quantite": [10, 7, 2, 1, 4, 17],
    "ville_epicerie": ["Tana", "Majunga", "Fianaratsoa", "Tana", "Majunga", "Manakara"],
})

#1
#print(df)
#df.info
#print(df.describe())

#2
df["ca"]= df["quantite"]* df["prixu"]
print(df)

#3
groupage1= df.groupby("ville_epicerie").agg(ca_ville= ("ca", "sum"))
print("Le CA par ville est :")
print(groupage1)
groupage2= df.groupby("categorie").agg(ca_categorie= ("ca", "sum"))
print("Le CA par catégorie est :")
print(groupage2)

#4
print("Les produits les plus vendus sont :")
p_vendu= df.nlargest(3,"quantite")
print(p_vendu)

#5
df["date"]= pd.to_datetime(df["date"], format = "mixed")
ca_month= df.set_index("date").resample("ME")["ca"].sum()
print(ca_month)

#6

df.fillna(df["quantite"].median())

#7
df.loc[3, "quantite"] = np.nan
croissement= pd.pivot_table(df, values="ca", index="categorie", columns="ville_epicerie", aggfunc="sum")
print(croissement)

#8
encoding= pd.get_dummies(df,columns=["categorie"], drop_first= True)
print(encoding)

#9 SQL
SELECT ville, SUM(ca) from df group by ville
SELECT categorie, SUM(ca) from df group by categorie13/18,  13 réponses corrects et 5 fausses lors du QCM
```
