```python
import pandas as pd

df = pd.DataFrame({
    "Produit": ["Ordinateur", "Téléphone", "Tablette", "Téléphone"],
    "Ville": ["Tana", "Majunga", "Tana", "Fianarantsoa"],
    "Prix": [2500, 800, None, 400],
    "Quantite": [3, 5, 2, 0]
})

print(df)

df.head() #affiche les 5 première lignes (5 est une valeur par défaut)
print(df.tail()) #affiche les 5 dernière lignes
print(df.shape)
print(df.describe()) #statistique rapide
print(df.columns) #affiche les étiquettes de colonnes
print("le nombre de ligne en double :",df.duplicated().sum()) #compte le nombre de ligne en doubles (doublons)
print("Le nombre de valeur manquante :",df.isnull().sum()) #compte le nombre de valeurs manquantes
print("Le nombre d'occurences des valeurs de la colonne 'Prix' :",df["Prix"].value_counts(normalize=False, dropna=False)) #compte le nbre d'occurences des valeurs de la colonne
print("Les valeurs uniques de la colonne 'Prix' :",df["Prix"].unique()) #affiche les valeurs unique (sans doublons)
print("Le nombre de valeurs uniques de la colonne 'Prix' :",df["Prix"].nunique()) #compte le nombre de valeurs unique (sans doublons)

df[df["Quantite"] > 2] #filtre les lignes où la quantité est supérieure à 2
df[df["Ville"].isin(["Tana", "Majunga"])] #filtre les lignes où la ville est soit "Tana" soit "Majunga"
df[df["Produit"].str.contains("téléphone", case=False)] #filtre les lignes où le produit contient la lettre "a" (insensible à la casse)
df.query("Quantite > 2 and Ville == 'Tana'") #filtre les lignes où la quantité est supérieure à 2 et la ville est "Tana"
```
