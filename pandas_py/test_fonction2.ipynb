```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "nom": ["Aina", "Hery", "Lalao", "Rina"],
    "age": [22, 30, 25, 20],
    "ville": ["Tana", "Tamatave", "Tana", "Majunga"]
})

groupage=df.groupby("ville")["age"].mean()
print(groupage)
df["age_dans_5_ans"] = df["age"] + 5
print(df)
df["majeur"] = np.where(df["age"] >= 25, "oui", "non") #comme sur excel , condition , valeur vrai , valeur faux
print(df["majeur"])
df.rename(columns={"nom": "prenom"}) #renommer les étiquettes de colonnes
print(df.columns)
df.drop(columns=["majeur"])
df.drop(index=[0, 1])
df.head()
df["age"] = df["age"].astype(float) #change le dtype
df["age"].apply(lambda x: x * 2) #applique la fonction lambda x à tout les élément de la colonnne cad multiplie tous les étéments à 2
df["ville"].map({"Tana": "Antananarivo", "Tamatave": "Fianaratsoa"}) #remplace les éléments contenus dans la liste par le nouveau nom , si ce n'est pas dans la liste = NaN
df["ville"].replace("Tana", "TNR") #remplace le nom de tous les Tana dans la colonne en TNR
```
