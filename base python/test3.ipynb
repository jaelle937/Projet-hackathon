```python
ma_bibliothèque= "Au coin deslivres"
jour_ouverture= ("Lundi", "Mercredi", "Vendredi", "Samedi")
print(jour_ouverture[0])
livres=[
    {"titre": "Le Petit Prince", "auteur": "Antoine de Saint-Exupéry", "page": 26, "emprunte": True},
    {"titre": "1984", "auteur": "George Orwell", "page": 109, "emprunte": True},
    {"titre": "To Kill a Mockingbird", "auteur": "Harper Lee", "page": 240, "emprunte": False},
    {"titre": "Pride and Prejudice", "auteur": "Jane Austen", "page": 81, "emprunte": True},
    {"titre": "The Great Gatsby", "auteur": "Harper Lee", "page": 92, "emprunte": False},
]

nbr_emprunte=0
for livre in livres:
    if livre["emprunte"]== True:
        nbr_emprunte += 1
        print("Le livre {} a été emprunté".format(livre["titre"]))
    else:
        print("Le livre {} est encore disponible".format(livre["titre"]))
print("Le nombre total de livre emprunté est de",nbr_emprunte)


def nbr_page(livres):
    pages = [livre["page"] for livre in livres]
    nbr_max = max(pages)
    nbr_min = min(pages)
    return nbr_max, nbr_min
nbr_page(livres)

def livres_de(livres, auteur):
    livres_auteur = [livre["titre"] for livre in livres if livre["auteur"] == auteur]
    return livres_auteur
print(livres_de(livres, "Harper Lee"))



def nombre_pages_moyen(livres):
    pages = [livre["page"] for livre in livres]
    return sum(pages) / len(pages)
print(nombre_pages_moyen(livres))

compteur = {}
for livre in livres:
    auteur = livre["auteur"]
    compteur[auteur] = compteur.get(auteur, 0) + 1
print(compteur)


class Livre:
    def __init__(self, titre, auteur, page):
        self.titre = titre
        self.auteur = auteur
        self.page = page
        self.emprunte = False

    def get_titre(self):
        return self.titre
    def get_auteur(self):
        return self.auteur
    def get_page(self):
        return self.page
    def est_disponible(self):
        return self.emprunte
    def emprunter(self):
        if self.emprunte:
            print("Le livre {} est déjà emprunté".format(self.titre))
        else:
            self.emprunte = True
            print("Le livre {} a été emprunté".format(self.titre))
    def rendre(self):
        if not self.emprunte:
            print("Le livre {} n'est pas emprunté".format(self.titre))
        else:
            self.emprunte = False
            print("Le livre {} est maintenant libre".format(self.titre))
livre1 = Livre("Verity", "Colleen Hoover", 302)
livre1.emprunter()
print(livre1.est_disponible())
livre1.rendre()
print(livre1.est_disponible())
```
