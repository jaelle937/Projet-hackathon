```python
mon_ecole= "Essca Antanimena"
promo=(2025,2026)
etudiants = [
    {"nom": "Moi", "note": 16},
    {"nom": "Eleanah", "note": 15},
    {"nom": "Luciano", "note": 17},
    {"nom": "Mickael", "note": 9},
    {"nom": "Rija", "note": 10},
]
for e in etudiants:
    if e["note"] >= 10:
        print(e["nom"], "est admis")
    else:
        print(e["nom"], "n'est pas admis")

total = 0
for e in etudiants:
    total += e["note"]
moyenne = total / len(etudiants)
print("La moyenne est de", moyenne)

class Etudiant:
    def __init__(self, nom, note):
        self.nom = nom
        self.note = note

    def get_nom(self):
        return self.nom

    def get_note(self):
        return self.note

    def est_admis(self):
        return self.note >= 10


e1 = Etudiant("Moi", 16)
e2 = Etudiant("Mickael", 9)

for e in [e1, e2]:
    if e.est_admis():
        print(e.get_nom(), "est admis")
    else:
        print(e.get_nom(), "n'est pas admis")
```
